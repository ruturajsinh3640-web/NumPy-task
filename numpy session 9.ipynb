```python
import numpy as np

prices = np.array([299, 499, 799, 0, 1599, -1, 899], dtype=float)
valid_mean = prices[prices > 0].mean()
prices[prices <= 0] = valid_mean

print("Updated prices:", prices)
print("Calculated mean:", valid_mean)
```

```python
import numpy as np

prices = np.array([299.0, 499.0, 799.0, 819.0, 1599.0, 819.0, 899.0])
quantities = np.array([2, 1, 3, 4, 2, 1, 5])
total_bills = prices * quantities

min_price = prices.min()
max_price = prices.max()
avg_price = prices.mean()
total_price_sum = prices.sum()
overall_total_bill = total_bills.sum()

print("--- Summary Report ---")
print(f"Minimum Cleaned Price : {min_price:.2f}")
print(f"Maximum Cleaned Price : {max_price:.2f}")
print(f"Average Cleaned Price : {avg_price:.2f}")
print(f"Sum of Unit Prices    : {total_price_sum:.2f}")
print(f"Total Bill (All Items): {overall_total_bill:.2f}")
```
