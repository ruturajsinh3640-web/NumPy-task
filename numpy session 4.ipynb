```python
import numpy as np

likes = np.array([120, 85, 210, 150, 95, 310, 175])
comments = np.array([14, 8, 25, 19, 11, 42, 22])

total_engagement = likes + comments

average_engagement = total_engagement.sum() / len(total_engagement)

print(f"Total engagement per post: {total_engagement}")
print(f"Average engagement per post: {average_engagement:.2f}")
```

```python
import numpy as np

prices = np.array([250, 420, 180, 310, 550])
discounts = np.array([30, 50, 20, 45, 75])

final_prices = prices - discounts

print("Final prices after discount:")
print(final_prices)
```

```python
import numpy as np

scores = np.array([172, 195, 180, 214, 168])

high_scoring_matches = scores > 180

print("Scores:", scores)
print("Scores greater than 180:", high_scoring_matches)
```

```python
import numpy as np

paytm = np.array([1, 0, 0, 1, 0, 1])
phonepe = np.array([0, 1, 0, 1, 1, 0])

paid_by_either = np.logical_or(paytm, phonepe)

print("Paytm:         ", paytm)
print("PhonePe:       ", phonepe)
print("Paid by either:", paid_by_either)
```

```python
import numpy as np

daily_steps = np.array([6200, 7800, 5400, 8900, 10200, 4500, 7100])

updated_steps = daily_steps + 500

total_steps = updated_steps.sum()

print("Original daily steps:", daily_steps)
print("Steps with 500 bonus:", updated_steps)
print("Total steps for the week:", total_steps)
```
