```python
import numpy as np

# 1. Initialize cart counters for 10 items to zero
cart = np.zeros(10, dtype=int)

# 2. Add the 3rd and 7th items to cart (using 0-based indexing: index 2 and index 6)
cart[2] = 1
cart[6] = 1

# 3. Print the updated cart state
print("Updated Cart Array:")
print(cart)
```

```python
import numpy as np

# 1. Set seed for reproducible results
np.random.seed(42)

# 2. Generate 6 random 4-digit OTP codes between 1000 and 9999 (inclusive)
# Note: high=10000 because the upper bound in np.random.randint() is exclusive
otp_codes = np.random.randint(low=1000, high=10000, size=6)

# 3. Print the generated array
print("Generated OTPs:")
print(otp_codes)
```
