```python
import time
import numpy as np

def compare_addition_speed(size=10_000):
    # Initialize the list and array
    py_list = list(range(size))
    np_arr = np.arange(size)

    # 1. Measure Python list addition (list comprehension)
    start_time = time.perf_counter()
    _ = [x + 5 for x in py_list]
    list_time = time.perf_counter() - start_time

    # 2. Measure NumPy array addition (vectorized broadcasting)
    start_time = time.perf_counter()
    _ = np_arr + 5
    array_time = time.perf_counter() - start_time

    # 3. Display results
    print(f"--- Addition Speed Comparison ({size:,} elements) ---")
    print(f"Python List: {list_time * 1000:.4f} ms")
    print(f"NumPy Array: {array_time * 1000:.4f} ms")
    
    speedup = list_time / array_time if array_time > 0 else float('inf')
    print(f"NumPy is ~{speedup:.1f}x faster")

if __name__ == "__main__":
    compare_addition_speed()
```

```python
import time
import numpy as np

# Sample Zomato-style ratings (scale 1.0 to 5.0) for 100,000 restaurant entries
sample_ratings = [3.8, 4.2, 4.9, 2.5, 3.1] * 20_000

# Convert to NumPy array
ratings_arr = np.array(sample_ratings)

start_loop = time.perf_counter()

doubled_ratings_loop = []
for rating in sample_ratings:
    doubled_ratings_loop.append(rating * 2)

loop_time = time.perf_counter() - start_loop


start_vec = time.perf_counter()

doubled_ratings_vec = ratings_arr * 2 

vec_time = time.perf_counter() - start_vec

print("Original Ratings (first 5):", sample_ratings[:5])
print("Doubled via Loop (first 5):", doubled_ratings_loop[:5])
print("Doubled via NumPy (first 5):", doubled_ratings_vec[:5])
print("\n--- Performance ---")
print(f"For-Loop Time:  {loop_time * 1000:.3f} ms")
print(f"NumPy Vec Time: {vec_time * 1000:.3f} ms")
print(f"Speedup:        {loop_time / vec_time:.1f}x faster")
```
